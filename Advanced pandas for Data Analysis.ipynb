## 1. Groupby

```python
import pandas as pd
```

```python
insurance_data = pd.read_csv("Insurance.csv")
insurance_data
```

### 1.1 Perform initial analysis

```python
insurance_data.isnull().sum()
```

```python
insurance_data.dtypes
```

```python
insurance_data.describe(include = "all")
```

### 1.2 Pull up Average Insurance Charges based on Region

```python
insurance_data.groupby(by = "region")["charges"].mean().round(2).sort_values()
```

### 1.3 Pull up Average Insurance Charges based on Region and Sex

```python
insurance_data.groupby(by = ["region","sex"])["charges"].mean().round(2).sort_values()
```

## 2. Pivot Table

```python
pd.pivot_table(data = insurance_data,values="charges",index="region",columns="sex").round(2)
```

## 3. Aggregate

### 3.1 Pull up Aggregate Functions for Insurance Charges based on Region and Sex

```python
insurance_data.groupby(by = ["region","sex"])["charges"].agg(["mean","max","min","sum"])
```

## 4. Cross Tab

### 4.1 How many Smokers and Non-Smokers are there based on Region

```python
insurance_data.head(5)
```

```python
pd.crosstab(index=insurance_data["region"],columns=insurance_data["smoker"],margins = True)
```

## 6. Apply

### 6.1 Create a new column in the same Dataframe with the name "Eligibility Criteria" where if age > 20, say Eligible or not

```python
def get_age(x) :
    if x > 20:
        return "Eligible"

    else:
        return "Not Eligible"


## If we want to update the existing dataframe, we need to use the return keyword
```

```python
insurance_data["Eligibilty"] = insurance_data["age"].apply(get_age)
insurance_data
```

## 7. Lambda Function

#### Anonymous Function / Nameless Function

```python
insurance_data.head(5)
```

```python
insurance_data["EC"] = insurance_data["age"].apply(lambda x : "Yes Eligible" if x > 20 else "No NE")
insurance_data
```

## Del keyword

```python
del insurance_data["EC"]
insurance_data
```
